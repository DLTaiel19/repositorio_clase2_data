# If

```python
x = 10
if x < 10:
    print("X es menor a 10")
elif x > 10:
    print("X es mayor a 10")
else:
    print("X es 10!") # ésta es la que saldrá en pantalla
```

# Case

```python
opcion = int(input("Ingrese un número (1-3): "))

match opcion:
    case 1:
        print("Elegiste la opción 1")
    case 2:
        print("Elegiste la opción 2")
    case 3:
        print("Elegiste la opción 3")
    case _:
        print("Opción inválida")
```
